```python
import os
import glob
import zipfile
import numpy as np
import pandas as pd

from google.colab import drive

# Mount Google Drive
drive.mount("/content/drive")

ZIP_PATH = "/content/drive/MyDrive/AI-Based-Vision-Systems.zip"
EXTRACT_DIR = "/content/AI-Based-Vision-Systems"

# Find existing feature file
matches = glob.glob(
    "/content/**/train_bovw_v1.npy",
    recursive=True
)

# Extract ZIP if necessary
if not matches and os.path.exists(ZIP_PATH):
    print("Extracting project ZIP...")

    with zipfile.ZipFile(ZIP_PATH, "r") as z:
        z.extractall(EXTRACT_DIR)

    matches = glob.glob(
        "/content/**/train_bovw_v1.npy",
        recursive=True
    )

if not matches:
    raise FileNotFoundError(
        "Feature file not found. Check the ZIP location in Google Drive."
    )

FEATURES = os.path.dirname(matches[0])

print("FEATURES =", FEATURES)

# Verify required files
required_files = [
    "train_bovw_v1.npy",
    "train_cnn_v1.npy",
    "train_hybrid_v1.npy",
    "val_bovw_v1.npy",
    "val_cnn_v1.npy",
    "val_hybrid_v1.npy",
    "train_labels_v1.npy",
    "val_labels_v1.npy",
]

for filename in required_files:
    path = os.path.join(FEATURES, filename)
    print(filename, ":", os.path.exists(path))
```

```python
import time
import random
import numpy as np
import pandas as pd
import joblib
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report
)
from tqdm.auto import tqdm

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("PyTorch:", torch.__version__)
print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("WARNING: GPU not detected. Enable T4 in Runtime settings.")
```

```python
import os
import numpy as np
import pandas as pd
# Load labels
y_train_raw = np.load(
    os.path.join(FEATURES, "train_labels_v1.npy")
)
y_val_raw = np.load(
    os.path.join(FEATURES, "val_labels_v1.npy")
)

# Encode labels consistently
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(y_train_raw)

# Check that validation classes exist in training classes
unknown_val_labels = set(y_val_raw) - set(y_train_raw)

assert not unknown_val_labels, (
    f"Validation contains unseen classes: {unknown_val_labels}"
)

y_val = label_encoder.transform(y_val_raw)

NUM_CLASSES = len(label_encoder.classes_)

print("Training samples:", len(y_train))
print("Validation samples:", len(y_val))
print("Classes:", NUM_CLASSES)

assert NUM_CLASSES == 256, (
    f"Expected 256 classes, found {NUM_CLASSES}"
)

# Load three representations
feature_sets = {
    "SIFT": {
        "train": np.load(os.path.join(FEATURES, "train_bovw_v1.npy")),
        "val": np.load(os.path.join(FEATURES, "val_bovw_v1.npy")),
        "expected_dim": 256,
    },
    "CNN": {
        "train": np.load(os.path.join(FEATURES, "train_cnn_v1.npy")),
        "val": np.load(os.path.join(FEATURES, "val_cnn_v1.npy")),
        "expected_dim": 512,
    },
    "Hybrid": {
        "train": np.load(os.path.join(FEATURES, "train_hybrid_v1.npy")),
        "val": np.load(os.path.join(FEATURES, "val_hybrid_v1.npy")),
        "expected_dim": 768,
    },
}

for name, data in feature_sets.items():
    Xtr = data["train"]
    Xva = data["val"]

    assert Xtr.ndim == 2 and Xva.ndim == 2
    assert Xtr.shape[0] == len(y_train)
    assert Xva.shape[0] == len(y_val)
    assert Xtr.shape[1] == data["expected_dim"]
    assert Xva.shape[1] == data["expected_dim"]
    assert np.isfinite(Xtr).all()
    assert np.isfinite(Xva).all()

    print(
        f"{name}: train={Xtr.shape}, validation={Xva.shape}"
    )
```

```python
class MLPClassifier(nn.Module):

    def __init__(self, input_dim, num_classes):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.30),

            nn.Linear(512, 256),
            nn.ReLU(),
            nn.Dropout(0.20),

            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        return self.network(x)


# Common training configuration
MAX_EPOCHS = 100
PATIENCE = 10
BATCH_SIZE = 256
LEARNING_RATE = 0.001

print("Maximum epochs:", MAX_EPOCHS)
print("Early stopping patience:", PATIENCE)
print("Batch size:", BATCH_SIZE)
print("Learning rate:", LEARNING_RATE)
```

```python
import os

# FEATURES variable pehle se defined hona chahiye
assert "FEATURES" in globals(), (
    "FEATURES missing hai. Pehle feature path setup karo."
)

PROJECT = os.path.dirname(os.path.dirname(FEATURES))

OUTPUT = os.path.join(
    PROJECT,
    "results",
    "fresh_colab_run",
    "pytorch_mlp_all"
)

MODEL_DIR = os.path.join(OUTPUT, "models")
REPORT_DIR = os.path.join(OUTPUT, "reports")

os.makedirs(MODEL_DIR, exist_ok=True)
os.makedirs(REPORT_DIR, exist_ok=True)

print("PROJECT:", PROJECT)
print("FEATURES:", FEATURES)
print("OUTPUT:", OUTPUT)
print("MODEL_DIR:", MODEL_DIR)
print("REPORT_DIR:", REPORT_DIR)
```

```python
all_results = []
all_histories = {}

for model_name, data in feature_sets.items():

    print("\n" + "=" * 70)
    print(f"TRAINING {model_name.upper()} MLP")
    print("=" * 70)

    start_time = time.time()

    # -------------------------
    # Feature scaling
    # -------------------------
    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(
        data["train"]
    ).astype(np.float32)

    X_val_scaled = scaler.transform(
        data["val"]
    ).astype(np.float32)

    Xtr_tensor = torch.from_numpy(X_train_scaled)
    Xva_tensor = torch.from_numpy(X_val_scaled)

    ytr_tensor = torch.tensor(y_train, dtype=torch.long)
    yva_tensor = torch.tensor(y_val, dtype=torch.long)

    train_loader = DataLoader(
        TensorDataset(Xtr_tensor, ytr_tensor),
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=2,
        pin_memory=(DEVICE.type == "cuda")
    )

    val_loader = DataLoader(
        TensorDataset(Xva_tensor, yva_tensor),
        batch_size=512,
        shuffle=False,
        num_workers=2,
        pin_memory=(DEVICE.type == "cuda")
    )

    # -------------------------
    # Model and optimizer
    # -------------------------
    model = MLPClassifier(
        input_dim=data["train"].shape[1],
        num_classes=NUM_CLASSES
    ).to(DEVICE)

    criterion = nn.CrossEntropyLoss()

    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=LEARNING_RATE
    )

    best_val_f1 = -1.0
    best_epoch = 0
    epochs_without_improvement = 0
    history = []

    model_path = os.path.join(
        MODEL_DIR, f"{model_name.lower()}_mlp_best.pth"
    )

    scaler_path = os.path.join(
        MODEL_DIR, f"{model_name.lower()}_mlp_scaler.joblib"
    )

    encoder_path = os.path.join(
        MODEL_DIR, f"{model_name.lower()}_mlp_label_encoder.joblib"
    )

    # -------------------------
    # Training loop
    # -------------------------
    for epoch in range(1, MAX_EPOCHS + 1):

        model.train()

        train_loss_sum = 0.0
        train_count = 0

        progress = tqdm(
            train_loader,
            desc=f"{model_name} | Epoch {epoch}/{MAX_EPOCHS}",
            leave=False
        )

        for xb, yb in progress:

            xb = xb.to(
                DEVICE,
                non_blocking=(DEVICE.type == "cuda")
            )
            yb = yb.to(
                DEVICE,
                non_blocking=(DEVICE.type == "cuda")
            )

            optimizer.zero_grad(set_to_none=True)

            logits = model(xb)
            loss = criterion(logits, yb)

            loss.backward()
            optimizer.step()

            n = xb.size(0)
            train_loss_sum += loss.item() * n
            train_count += n

            progress.set_postfix(
                loss=f"{loss.item():.4f}"
            )

        train_loss = train_loss_sum / train_count

        # -------------------------
        # Validation
        # -------------------------
        model.eval()

        val_loss_sum = 0.0
        val_count = 0
        epoch_true = []
        epoch_pred = []

        with torch.inference_mode():

            for xb, yb in val_loader:

                xb = xb.to(
                    DEVICE,
                    non_blocking=(DEVICE.type == "cuda")
                )
                yb = yb.to(
                    DEVICE,
                    non_blocking=(DEVICE.type == "cuda")
                )

                logits = model(xb)
                loss = criterion(logits, yb)

                pred = logits.argmax(dim=1)

                n = xb.size(0)
                val_loss_sum += loss.item() * n
                val_count += n

                epoch_true.extend(yb.cpu().numpy())
                epoch_pred.extend(pred.cpu().numpy())

        val_loss = val_loss_sum / val_count

        val_accuracy = accuracy_score(
            epoch_true, epoch_pred
        )

        val_macro_f1 = f1_score(
            epoch_true,
            epoch_pred,
            average="macro",
            zero_division=0
        )

        history.append({
            "epoch": epoch,
            "train_loss": train_loss,
            "val_loss": val_loss,
            "val_accuracy": val_accuracy,
            "val_macro_f1": val_macro_f1
        })

        # -------------------------
        # Save best checkpoint
        # -------------------------
        if val_macro_f1 > best_val_f1:

            best_val_f1 = val_macro_f1
            best_epoch = epoch
            epochs_without_improvement = 0

            torch.save(
                {
                    "model_name": model_name,
                    "epoch": epoch,
                    "input_dim": data["train"].shape[1],
                    "num_classes": NUM_CLASSES,
                    "model_state_dict": model.state_dict(),
                    "val_accuracy": val_accuracy,
                    "val_macro_f1": val_macro_f1,
                    "val_loss": val_loss
                },
                model_path
            )

            joblib.dump(scaler, scaler_path)
            joblib.dump(label_encoder, encoder_path)

            status = "BEST CHECKPOINT SAVED"

        else:
            epochs_without_improvement += 1
            status = (
                f"No improvement "
                f"{epochs_without_improvement}/{PATIENCE}"
            )

        print(
            f"{model_name} | Epoch {epoch:03d} | "
            f"Train Loss {train_loss:.4f} | "
            f"Val Loss {val_loss:.4f} | "
            f"Accuracy {val_accuracy * 100:.2f}% | "
            f"Macro-F1 {val_macro_f1 * 100:.2f}% | "
            f"{status}"
        )

        if epochs_without_improvement >= PATIENCE:
            print(
                f"Early stopping. Best epoch: {best_epoch}"
            )
            break

    training_seconds = time.time() - start_time
    all_histories[model_name] = pd.DataFrame(history)

    # -------------------------
    # Evaluate saved best model
    # -------------------------
    checkpoint = torch.load(
        model_path,
        map_location=DEVICE,
        weights_only=True
    )

    best_model = MLPClassifier(
        input_dim=checkpoint["input_dim"],
        num_classes=checkpoint["num_classes"]
    ).to(DEVICE)

    best_model.load_state_dict(
        checkpoint["model_state_dict"]
    )
    best_model.eval()

    final_true = []
    final_pred = []

    with torch.inference_mode():

        for xb, yb in val_loader:

            xb = xb.to(
                DEVICE,
                non_blocking=(DEVICE.type == "cuda")
            )

            pred = best_model(xb).argmax(dim=1)

            final_true.extend(yb.numpy())
            final_pred.extend(pred.cpu().numpy())

    final_accuracy = accuracy_score(
        final_true, final_pred
    )

    final_macro_f1 = f1_score(
        final_true,
        final_pred,
        average="macro",
        zero_division=0
    )

    # Classification report
    report = classification_report(
        final_true,
        final_pred,
        labels=np.arange(NUM_CLASSES),
        target_names=label_encoder.classes_,
        output_dict=True,
        zero_division=0
    )

    pd.DataFrame(report).transpose().to_csv(
        os.path.join(
            REPORT_DIR,
            f"{model_name.lower()}_mlp_report.csv"
        )
    )

    # Validation predictions
    predictions_df = pd.DataFrame({
        "true_label": label_encoder.inverse_transform(
            np.asarray(final_true)
        ),
        "predicted_label": label_encoder.inverse_transform(
            np.asarray(final_pred)
        )
    })

    predictions_df.to_csv(
        os.path.join(
            REPORT_DIR,
            f"{model_name.lower()}_mlp_predictions.csv"
        ),
        index=False
    )

    # Training history
    all_histories[model_name].to_csv(
        os.path.join(
            REPORT_DIR,
            f"{model_name.lower()}_mlp_history.csv"
        ),
        index=False
    )

    result = {
        "model": f"{model_name} + PyTorch MLP",
        "input_dimensions": data["train"].shape[1],
        "best_epoch": checkpoint["epoch"],
        "accuracy": final_accuracy,
        "macro_f1": final_macro_f1,
        "training_time_minutes": training_seconds / 60,
        "checkpoint": model_path
    }

    all_results.append(result)

    print(f"\n{model_name} MLP finished.")
    print(f"Best epoch: {checkpoint['epoch']}")
    print(f"Validation accuracy: {final_accuracy * 100:.2f}%")
    print(f"Validation Macro-F1: {final_macro_f1 * 100:.2f}%")
    print(f"Training time: {training_seconds / 60:.2f} minutes")
    print(f"Saved model: {model_path}")

print("\nALL THREE MLP EXPERIMENTS FINISHED.")
```

```python
results_df = pd.DataFrame(all_results)

results_df = results_df.sort_values(
    "macro_f1",
    ascending=False
).reset_index(drop=True)

results_df["accuracy_percent"] = (
    results_df["accuracy"] * 100
).round(2)

results_df["macro_f1_percent"] = (
    results_df["macro_f1"] * 100
).round(2)

results_df["training_time_minutes"] = (
    results_df["training_time_minutes"]
).round(2)

display(
    results_df[
        [
            "model",
            "input_dimensions",
            "best_epoch",
            "accuracy_percent",
            "macro_f1_percent",
            "training_time_minutes"
        ]
    ]
)

comparison_path = os.path.join(
    OUTPUT, "mlp_comparison.csv"
)

results_df.to_csv(comparison_path, index=False)

print("Comparison saved:", comparison_path)
```
